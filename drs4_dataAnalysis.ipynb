```python
import uproot
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

fname = "drs4test.root"

with uproot.open(fname) as f:
    df = f["events"].arrays(library="pd")

# tempo: unixtime è in secondi, con i ms
df["timestamp"] = pd.to_datetime(df["unixtime"], unit="s")
df["time_in_seconds"] = df["unixtime"] - df["unixtime"].min()

print(df.shape)
df.head()
```

```python
df["ch1_amp"].hist(bins=100)
plt.xlabel("Amplitude [V]"); 
plt.ylabel("Events")
plt.grid(False)
```

```python
plt.hist(df["ch1_charge"], bins=100, color = "orange")
plt.xlabel("Charge [C]"); 
plt.ylabel("Events")
plt.grid(False)
```
